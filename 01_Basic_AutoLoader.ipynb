```python
source_path = "/Volumes/mythri_databricks/etl_demo/raw_files/"

df = (
    spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "csv")
        .option(
            "cloudFiles.schemaLocation",
            "/Volumes/mythri_databricks/etl_demo/raw_files/_schema/customers"
        )
        .option("header", "true")
        .load(source_path)
)
```

```python
(
    df.writeStream
      .option(
          "checkpointLocation",
          "/Volumes/mythri_databricks/etl_demo/raw_files/_checkpoints/customers"
      )
      .trigger(availableNow=True)
      .toTable(
          "mythri_databricks.etl_demo.customers_autoloader_bronze"
      )
)
```
