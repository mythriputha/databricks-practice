```python
source_path = "abfss://different-data@mythriazurelearning.dfs.core.windows.net/"

df = (
    spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", "/Volumes/mythri_databricks/sales_bronze/sales_input/_schema/customers")
    .option("multiLine", "true")
    .load(source_path)
)

(
    df.writeStream
    .option("checkpointLocation", "/Volumes/mythri_databricks/sales_bronze/sales_input/_checkpoints/customers")
    .trigger(availableNow=True)
    .toTable("mythri_databricks.sales_bronze.customers_bronze")
)
```
