```python
from pyspark.sql import functions as F

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/mythri_databricks/sales_bronze/sales_input/sales.csv")
)

df.show()
```

```python
bronze_df = df.withColumn(
    "_ingest_ts",
    F.current_timestamp()
)
```

```python
bronze_df.show()
```

```python
bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "mythri_databricks.sales_bronze.raw_sales"
    )
```

```python
df = spark.sql("""
    SELECT *
    FROM mythri_databricks.sales_bronze.raw_sales
""")

df.show()
```

```python
silver_df = df.dropDuplicates(["SaleID"])
```

```python
silver_df.show()
```

```python
silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "mythri_databricks.sales_silver.cleaned_sales"
    )
```

```python
%sql

SELECT *
FROM mythri_databricks.sales_silver.cleaned_sales;
```

```python
silver_df = spark.sql("""
    SELECT *
    FROM mythri_databricks.sales_silver.cleaned_sales
""")

silver_df.show()
```

```python
from pyspark.sql import functions as F

region_gold = (
    silver_df
    .groupBy("Region")
    .agg(
        F.countDistinct("SaleID").alias("SaleCount"),
        F.sum("SalesAmount").alias("TotalSales"),
        F.avg("SalesAmount").alias("AverageSale"),
        F.max("SalesAmount").alias("HighestSale"),
        F.min("SalesAmount").alias("LowestSale")
    )
    .orderBy(
        F.col("TotalSales").desc()
    )
)
```

```python
region_gold.show()
```

```python
region_gold.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "mythri_databricks.sales_gold.region_sales_summary"
    )
```
