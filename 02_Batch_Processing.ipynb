```python
df = spark.read.table(
    "mythri_databricks.sales_bronze.customers_bronze"
)

display(df)
```

```python
from pyspark.sql import functions as F

silver_df = (
    df
    .dropDuplicates(["customer_id"])
    .withColumn(
        "purchase_amount",
        F.col("purchase_amount").cast("double")
    )
)

display(silver_df)
```

```python
silver_df.write.mode("overwrite").saveAsTable(
    "mythri_databricks.sales_silver.customers_silver"
)
```

```python
display(
    spark.sql("""
        SELECT *
        FROM mythri_databricks.sales_silver.customers_silver
    """)
)
```
