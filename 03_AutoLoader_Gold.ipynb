```python
silver_df = spark.read.table(
    "mythri_databricks.sales_silver.customers_silver"
)

display(silver_df)
```

```python
from pyspark.sql.functions import sum

gold_df = (
    silver_df
    .groupBy("city")
    .agg(
        sum("purchase_amount").alias("total_purchase_amount")
    )
)

display(gold_df)
```

```python
gold_df.write.mode("overwrite").saveAsTable(
    "mythri_databricks.sales_gold.customers_gold"
)
```
