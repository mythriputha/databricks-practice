```python
from pyspark.sql import functions as F

df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/mythri_databricks/sales_bronze/sales_input/sales.csv")
)

df.show()
```

```python
bronze_df = df.withColumn(
    "_ingest_ts",
    F.current_timestamp()
)
```

```python
bronze_df.show()
```

```python
bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "mythri_databricks.sales_bronze.raw_sales"
    )
```

```python
%sql
SELECT *
FROM mythri_databricks.sales_bronze.raw_sales;
```
